```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import LabelEncoder
```

```python
df=pd.read_csv("sales.csv")
```

```python
print(df.head())
```

```python
encoder=LabelEncoder()
df['loyalty_encoded']=encoder.fit_transform(df['loyalty_tier'])
```

```python
print(df[['loyalty_tier','loyalty_encoded']].head(20))
```

```python
x=df[['age','loyalty_encoded']]
```

```python
model=KMeans(n_clusters=3,
             random_state=42)
model.fit(x)
```

```python
df['clusters']=model.labels_
print("Clustered customers")
```

```python
print(model.cluster_centers_)
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df['age'],
    df['loyalty_encoded'],
    c=df['clusters'],
    cmap='rainbow'
)

plt.scatter(
    model.cluster_centers_[:,0],
    model.cluster_centers_[:,1],
    s=200,
    marker='X'
)

plt.xlabel("Age")
plt.ylabel("Loyalty Tier")
plt.title("Customer Segmentation")
plt.yticks([0,1,2], ['Bronze','Gold','Silver'])

plt.show()
```
